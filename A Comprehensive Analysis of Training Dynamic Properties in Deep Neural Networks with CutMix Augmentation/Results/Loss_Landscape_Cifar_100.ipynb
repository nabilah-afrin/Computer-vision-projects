```python
# Function to plot loss landscapes
def plot_loss_landscapes(histories_cutmix, histories, epochs=50):
    losses_bn = {lr: np.array(hist["loss"]) for lr, hist in histories_cutmix.items()}
    losses_no_bn = {lr: np.array(hist["loss"]) for lr, hist in histories.items()}

    min_length_bn = min(len(loss) for loss in losses_bn.values())
    min_length_no_bn = min(len(loss) for loss in losses_no_bn.values())
    min_length = min(min_length_bn, min_length_no_bn)

    min_curve_bn, max_curve_bn, mean_curve_bn = [], [], []
    min_curve_no_bn, max_curve_no_bn, mean_curve_no_bn = [], [], []

    for i in range(epochs):
        loss_values_at_epoch_bn = [loss[i] for loss in losses_bn.values() if i < len(loss) and not np.isnan(loss[i])]
        if loss_values_at_epoch_bn:  # Ensure non-empty list
            min_curve_bn.append(np.min(loss_values_at_epoch_bn))
            max_curve_bn.append(np.max(loss_values_at_epoch_bn))
            # mean_curve_bn.append(np.mean(loss_values_at_epoch_bn))

    for i in range(epochs):
        loss_values_at_epoch_no_bn = [loss[i] for loss in losses_no_bn.values() if i < len(loss) and not np.isnan(loss[i])]
        if loss_values_at_epoch_no_bn:  # Ensure non-empty list
            min_curve_no_bn.append(np.min(loss_values_at_epoch_no_bn))
            max_curve_no_bn.append(np.max(loss_values_at_epoch_no_bn))
            # mean_curve_no_bn.append(np.mean(loss_values_at_epoch_no_bn))

    plt.figure(figsize=(15, 10))

    plt.fill_between(range(len(min_curve_bn)), min_curve_bn, max_curve_bn, color='green', alpha=0.5, label='ConvMixer')
    # plt.plot(mean_curve_bn, color='green', linestyle='-', label='Mean loss (ConvMixer)')
    for lr, losses in losses_bn.items():
        plt.plot(losses[:min_length], label=f'ConvMixer lr={lr}', linestyle='--')



    plt.fill_between(range(len(min_curve_no_bn)), min_curve_no_bn, max_curve_no_bn, color='red', alpha=0.5, label='ResNet')
    # plt.plot(mean_curve_no_bn, color='red', linestyle='-', label='Mean loss (ResNet)')
    for lr, losses in losses_no_bn.items():
        plt.plot(losses[:min_length], label=f'ResNet lr={lr}', linestyle='-.')

    plt.xlabel('Epochs', fontsize=9)
    plt.ylabel('Loss', fontsize=9)
    plt.title('Training Loss Landscapes with Standard Augmentation at Different Learning Rates: CIFAR-100')
    plt.legend(loc='best', fontsize=9)
    plt.grid(True)
    plt.show()

# Plot the loss landscapes
plot_loss_landscapes(histories_cutmix, histories, epochs=50)
```
